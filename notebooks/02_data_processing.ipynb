```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

### 1. Load Cleaned Data and Normalize Pollutant Values
Load the data from `cleaned_air_quality.csv`. Min-Max scaling is then applied to the pollutant columns to normalize their values between 0 and 1.

```python
# Load the cleaned data
final_df = pd.read_csv('cleaned_air_quality.csv', low_memory=False)

# Define pollutant columns and initialize the scaler
pollutant_columns = ['pm2.5_avg', 'pm10_avg', 'so2_avg']
scaler = MinMaxScaler()

# Separate rows that have valid pollutant data for scaling
valid_rows = final_df.dropna(subset=pollutant_columns)

# Apply Min-Max scaling
if not valid_rows.empty:
    scaled_values = scaler.fit_transform(valid_rows[pollutant_columns])
    final_df.loc[valid_rows.index, pollutant_columns] = scaled_values
```

### 2. Feature Engineering and Final Export
The `date` column is converted to a datetime object, and the `year` is extracted into a new column for easier time-based analysis. The fully processed DataFrame is then saved.

```python
# Convert the 'date' column to datetime objects
final_df['date'] = pd.to_datetime(final_df['date'], errors='coerce')

# Extract the year to create a new 'year' column
final_df['year'] = final_df['date'].dt.year

# Save the fully processed data to a new CSV file
final_df.to_csv('processed_air_quality_data.csv', index=False)
```
