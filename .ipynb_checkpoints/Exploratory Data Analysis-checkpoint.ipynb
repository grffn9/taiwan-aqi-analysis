```python
# Code from data processing file

import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Define the file name
file_name = 'air_quality.csv'

# Set the chunk size, this defines how many rows you want to load at once
chunk_size = 500000  # You can adjust this based on your system's memory

# Create an empty list to store the chunks
chunks = []

# Use the pandas read_csv function to read the file in chunks
for chunk in pd.read_csv(file_name, chunksize=chunk_size, low_memory=False):
    # Perform any operations on each chunk (e.g., drop columns or rows with missing data)
    chunks.append(chunk)

# Drop the 'unit' column since it has too many missing values
for i, chunk in enumerate(chunks):
    if 'unit' in chunk.columns:
        chunks[i] = chunk.drop(columns=['unit'])

# Remove rows where the 'aqi' column is missing
for i, chunk in enumerate(chunks):
    chunks[i] = chunk.dropna(subset=['aqi'])

# Convert Date Column
for i, chunk in enumerate(chunks):
    chunks[i]['date'] = pd.to_datetime(chunks[i]['date'], errors='coerce')

# List of pollutant columns to normalize
pollutant_columns = ['pm2.5_avg', 'pm10_avg', 'so2_avg']

# Initialize the Min-Max scaler
scaler = MinMaxScaler()

# Process data in chunks
chunks = []  # To store processed chunks

for chunk in pd.read_csv(file_name, chunksize=chunk_size, low_memory=False):
    # Step 1: Replace '-' with NaN in the pollutant columns
    for col in pollutant_columns:
        chunk[col] = chunk[col].replace('-', np.nan)
    
    # Step 2: Convert columns to numeric (this will ensure any remaining non-numeric values are converted to NaN)
    chunk[pollutant_columns] = chunk[pollutant_columns].apply(pd.to_numeric, errors='coerce')
    
    # Step 3: Identify rows without NaN in pollutant columns
    valid_rows = chunk.dropna(subset=pollutant_columns)
    
    # Step 4: Apply Min-Max scaling to the valid rows (leaving NaN rows untouched)
    if not valid_rows.empty:
        scaled_values = scaler.fit_transform(valid_rows[pollutant_columns])
        chunk.loc[valid_rows.index, pollutant_columns] = scaled_values

    # Append the chunk (with both scaled and NaN rows) to the list
    chunks.append(chunk)

# Concatenate all processed chunks back into a single DataFrame if needed
final_df = pd.concat(chunks, ignore_index=True)

final_df.head()
```

```python
# Adjusting the display format for better readability
pd.options.display.float_format = '{:.2f}'.format

# Display the summary statistics again
print(final_df[pollutant_columns].describe())
final_df.shape
```

```python
import matplotlib.pyplot as plt

# Create a figure with 3 subplots, one for each pollutant
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Plot histograms for each pollutant, with a logarithmic y-axis scale
final_df['pm2.5_avg'].plot(kind='hist', bins=100, ax=axes[0], log=True, color='skyblue', edgecolor='black')
axes[0].set_title('PM2.5 Distribution')
axes[0].set_xlabel('PM2.5 (μg/m³)')
axes[0].set_ylabel('Frequency (log scale)')

final_df['pm10_avg'].plot(kind='hist', bins=100, ax=axes[1], log=True, color='orange', edgecolor='black')
axes[1].set_title('PM10 Distribution')
axes[1].set_xlabel('PM10 (μg/m³)')
axes[1].set_ylabel('Frequency (log scale)')

final_df['so2_avg'].plot(kind='hist', bins=100, ax=axes[2], log=True, color='green', edgecolor='black')
axes[2].set_title('SO2 Distribution')
axes[2].set_xlabel('SO2 (ppb)')
axes[2].set_ylabel('Frequency (log scale)')

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()
```

```python
# Create subplots for the rolling mean to observe trends
fig, axes = plt.subplots(3, 1, figsize=(15, 12))

# Rolling mean window size (e.g., 30 days)
window_size = 30

# PM2.5 rolling mean
final_df.set_index('date')['pm2.5_avg'].rolling(window=window_size).mean().plot(ax=axes[0], color='blue', linewidth=2)
axes[0].set_title(f'PM2.5 Levels Over Time (Rolling Mean, {window_size} Days)')
axes[0].set_ylabel('PM2.5 (μg/m³)')
axes[0].set_xlabel('Date')

# PM10 rolling mean
final_df.set_index('date')['pm10_avg'].rolling(window=window_size).mean().plot(ax=axes[1], color='orange', linewidth=2)
axes[1].set_title(f'PM10 Levels Over Time (Rolling Mean, {window_size} Days)')
axes[1].set_ylabel('PM10 (μg/m³)')
axes[1].set_xlabel('Date')

# SO2 rolling mean
final_df.set_index('date')['so2_avg'].rolling(window=window_size).mean().plot(ax=axes[2], color='green', linewidth=2)
axes[2].set_title(f'SO2 Levels Over Time (Rolling Mean, {window_size} Days)')
axes[2].set_ylabel('SO2 (ppb)')
axes[2].set_xlabel('Date')

# Adjust layout and display the plots
plt.tight_layout()
plt.show()
```

```python
'#9ACD32'

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Define the file name
file_name = 'air_quality.csv'

# Set the chunk size, this defines how many rows you want to load at once
chunk_size = 500000  # You can adjust this based on your system's memory

# List of pollutant columns (No Scaling will be applied)
pollutant_columns = ['pm2.5_avg', 'pm10_avg', 'so2_avg']

# Create an empty list to store the cleaned chunks
processed_chunks = []

# Use the pandas read_csv function to read the file in chunks
for chunk in pd.read_csv(file_name, chunksize=chunk_size, low_memory=False):
    
    # Drop the 'unit' column since it has too many missing values
    if 'unit' in chunk.columns:
        chunk = chunk.drop(columns=['unit'])

    # Remove rows where the 'aqi' column is missing
    chunk = chunk.dropna(subset=['aqi'])

    # Convert the 'date' column to datetime, handling known format: "%Y-%m-%d %H:%M"
    chunk['date'] = pd.to_datetime(chunk['date'], format="%Y-%m-%d %H:%M", errors='coerce')

    # Replace '-' with NaN in pollutant columns
    for col in pollutant_columns:
        chunk[col] = chunk[col].replace('-', np.nan)

    # Convert pollutant columns to numeric, replacing non-numeric values with NaN
    chunk[pollutant_columns] = chunk[pollutant_columns].apply(pd.to_numeric, errors='coerce')

    # Append the cleaned chunk to the list
    processed_chunks.append(chunk)

# Concatenate all processed chunks back into a single DataFrame
final_df = pd.concat(processed_chunks, ignore_index=True)

# Drop rows with invalid dates
final_df_cleaned = final_df.dropna(subset=['date'])

# Extract the year from the 'date' column
final_df_cleaned['year'] = final_df_cleaned['date'].dt.year

# Group by year and calculate the average PM10 concentration
yearly_avg_pm10 = final_df_cleaned.groupby('year')['pm10_avg'].mean()

# Verify the unique years in the data
print("Unique years in the data after cleaning:", final_df_cleaned['year'].unique())

# Plot the average PM10 concentration by year
plt.figure(figsize=(10, 6))
yearly_avg_pm10.plot(kind='bar', color='green', edgecolor='black')

# Add labels and title
plt.title('Average PM10 Concentration by Year (2016-2024)')
plt.xlabel('Year')
plt.ylabel('Average PM10 Concentration (μg/m³)')

# Display the plot
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Plot the average PM10 concentration by year
plt.figure(figsize=(12, 7))

# Create the bar plot with enhanced aesthetics and smaller bar width
bars = plt.bar(yearly_avg_pm10.index, yearly_avg_pm10.values, width=0.6, color='#8FBF73', edgecolor='black')

# Add value labels on top of the bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval, round(yval, 1), ha='center', va='bottom', fontsize=10)

# Ensure all x-axis labels are shown without alternation and reduce rotation
plt.xticks(yearly_avg_pm10.index, rotation=0, fontsize=12)

# Customize the y-axis label and title
plt.ylabel('Average PM10 Concentration (μg/m³)', fontsize=14)
plt.xlabel('Year', fontsize=14)
plt.title('Average PM10 Concentration by Year (2016-2024)', fontsize=16)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Define a colormap that transitions from red (high values) to green (low values)
cmap = plt.get_cmap('RdYlGn_r')  # 'r' reverses it to have red for high values

# Normalize the PM10 values for mapping to the color gradient
norm = plt.Normalize(yearly_avg_pm10.min(), yearly_avg_pm10.max())

# Plot the average PM10 concentration by year
plt.figure(figsize=(12, 7))

# Create the bar plot using the gradient colormap based on PM10 values
bars = plt.bar(yearly_avg_pm10.index, yearly_avg_pm10.values, color=cmap(norm(yearly_avg_pm10.values)), edgecolor='black')

# Add value labels on top of the bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval, round(yval, 1), ha='center', va='bottom', fontsize=10)

# Add gridlines for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Ensure all x-axis labels are shown clearly and adjust font size
plt.xticks(yearly_avg_pm10.index, rotation=0, fontsize=12)

# Customize the y-axis label and title
plt.ylabel('Average PM10 Concentration (μg/m³)', fontsize=14)
plt.xlabel('Year', fontsize=14)
plt.title('Average PM10 Concentration by Year (2016-2024)', fontsize=16)

# Add a colorbar legend to show the gradient from dangerous to safe levels
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])  # Only needed for the colorbar
cbar = plt.colorbar(sm)
cbar.set_label('PM10 Concentration (μg/m³)', fontsize=12)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
# Ensure the 'date' column is in datetime format
final_df['date'] = pd.to_datetime(final_df['date'], errors='coerce')

# Now extract month from the date
final_df['month'] = final_df['date'].dt.month

# Create a single plot for all seasonal variations
plt.figure(figsize=(15, 8))

# PM10 seasonal variation (color close to brown)
final_df.groupby('month')['pm10_avg'].mean().plot(label='PM10', color='#8B4513', linewidth=2)

# PM2.5 seasonal variation (color close to blue)
final_df.groupby('month')['pm2.5_avg'].mean().plot(label='PM2.5', color='#3366CC', linewidth=2)

# SO2 seasonal variation (color close to green)
final_df.groupby('month')['so2_avg'].mean().plot(label='SO2', color='#9ACD32', linewidth=2)

# Set title and labels
plt.title('Seasonal Variations of Air Quality', fontsize=16)
plt.ylabel('Concentration by Pollutant', fontsize=12)
plt.xlabel('Month', fontsize=12)

# Add a legend
plt.legend(title='Air Quality Parameter', fontsize=12)

# Show plot
plt.tight_layout()
plt.show()
```

```python
# Create a figure for box plots to visualize outliers
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# PM2.5 outliers (box plot)
final_df.boxplot(column='pm2.5_avg', ax=axes[0], patch_artist=True, boxprops=dict(facecolor='skyblue'))
axes[0].set_title('PM2.5 Levels (Box Plot)')
axes[0].set_ylabel('PM2.5 (μg/m³)')

# PM10 outliers (box plot)
final_df.boxplot(column='pm10_avg', ax=axes[1], patch_artist=True, boxprops=dict(facecolor='orange'))
axes[1].set_title('PM10 Levels (Box Plot)')
axes[1].set_ylabel('PM10 (μg/m³)')

# SO2 outliers (box plot)
final_df.boxplot(column='so2_avg', ax=axes[2], patch_artist=True, boxprops=dict(facecolor='lightgreen'))
axes[2].set_title('SO2 Levels (Box Plot)')
axes[2].set_ylabel('SO2 (ppb)')

# Adjust layout and display the plots
plt.tight_layout()
plt.show()
```
