```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Define the file name
file_name = 'air_quality.csv'

# Set the chunk size, this defines how many rows you want to load at once
chunk_size = 500000  # You can adjust this based on your system's memory

# Create an empty list to store the chunks
chunks = []

# Use the pandas read_csv function to read the file in chunks
for chunk in pd.read_csv(file_name, chunksize=chunk_size, low_memory=False):
    # Perform any operations on each chunk (e.g., drop columns or rows with missing data)
    chunks.append(chunk)

# Display the first few rows of one of the chunks
chunks[0].head()
```

```python
# Check for missing values in the first chunk
missing_values = chunks[0].isnull().sum()

# Display the columns with missing values and their count
print(missing_values[missing_values > 0])
```

```python
# Drop the 'unit' column since it has too many missing values
for i, chunk in enumerate(chunks):
    if 'unit' in chunk.columns:
        chunks[i] = chunk.drop(columns=['unit'])
```

```python
# Remove rows where the 'aqi' column is missing
for i, chunk in enumerate(chunks):
    chunks[i] = chunk.dropna(subset=['aqi'])
```

```python
# Convert Date Column
for i, chunk in enumerate(chunks):
    chunks[i]['date'] = pd.to_datetime(chunks[i]['date'], errors='coerce')
```

```python
# List of pollutant columns to normalize
pollutant_columns = ['pm2.5_avg', 'pm10_avg', 'so2_avg']

# Initialize the Min-Max scaler
scaler = MinMaxScaler()

# Process data in chunks
chunks = []  # To store processed chunks

for chunk in pd.read_csv(file_name, chunksize=chunk_size, low_memory=False):
    # Step 1: Replace '-' with NaN in the pollutant columns
    for col in pollutant_columns:
        chunk[col] = chunk[col].replace('-', np.nan)
    
    # Step 2: Convert columns to numeric (this will ensure any remaining non-numeric values are converted to NaN)
    chunk[pollutant_columns] = chunk[pollutant_columns].apply(pd.to_numeric, errors='coerce')
    
    # Step 3: Identify rows without NaN in pollutant columns
    valid_rows = chunk.dropna(subset=pollutant_columns)
    
    # Step 4: Apply Min-Max scaling to the valid rows (leaving NaN rows untouched)
    if not valid_rows.empty:
        scaled_values = scaler.fit_transform(valid_rows[pollutant_columns])
        chunk.loc[valid_rows.index, pollutant_columns] = scaled_values

    # Append the chunk (with both scaled and NaN rows) to the list
    chunks.append(chunk)

# Concatenate all processed chunks back into a single DataFrame if needed
final_df = pd.concat(chunks, ignore_index=True)

final_df.head()
```
